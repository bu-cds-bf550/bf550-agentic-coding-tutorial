## Part 3 · Design your tutor

**In this part:** you can work out a specification with Claude Code before anything is built, by letting it interview you, and you can tell the decisions you made from the ones it made for you.

For the rest of today you are building one thing: **a personal tutor for BF550**, as a Claude Code skill. Its job is to help *you* understand this course's material better. Not a generic tutor and not the one your neighbour would want: one built around what you think is the way you learn best. You design it, with Claude, and Claude builds it.

The work follows the workflow from the slides, *explore, plan, code, commit*, in five steps:

| Step | What you do | What you have at the end |
|---|---|---|
| 1. Think (3.1) | notes on how you learn, without Claude | a few lines of notes |
| 2. Interview (3.3) | Claude asks, one question at a time; you answer | decisions, in the conversation |

| 4. Build (4.2, 4.3) | Claude plans the skill from your spec, then writes it | `SKILL.md`, your tutor |
| 5. Test (4.5) | you try your spec's three tests and revise | a tutor that does what your spec says |

### 3.1 · Step 1: how do you learn?

Take three minutes on your own, without Claude. Think back over this course so far: a moment when something clicked, and a moment when it didn't. What made the difference? Some things that help people, though maybe not you:

- being quizzed, and getting it wrong before getting it right;
- a worked example first, then one to do yourself;
- predicting what code prints before running it;
- explaining an idea back in your own words;
- a picture, an analogy, or the biology behind the statistics;
- short sessions often, or one long one;
- someone who is blunt about mistakes, or someone who is gentle.

### 3.2 · What a spec looks like

Your spec will be a markdown file with these sections. You don't fill them in yourself: the interview fills them, and then you edit.

```markdown
# My BF550 tutor

## Goal
## How I learn best
## Starting a session
## What it reads before it teaches
## How it teaches
## When I get something wrong
## Pace, length and ending
## Topics to come back to
## How I'll know it works
1. When I say "...", it should ...
2. When I say "...", it should ...
3. When I say "...", it should ...
```

Claude builds from what the spec *says*, so each line has to be specific enough to act on. A line is specific enough when you could watch the tutor and say whether it did it:

| Too vague to build | Specific enough to build |
|---|---|
| Quiz me well. | Ask one question at a time, and wait for my answer before the next. |
| Be encouraging. | When I'm wrong, say first what I got right, then the mistake in one sentence. |
| Use the course material. | Before a quiz, read the unit page for the topic and use its notation. |
| Not too long. | Stop after ten questions or when I say stop, and list what to review, with links. |
| Make it hard. | If I get two right in a row, make the next one harder. |

## Part 4 · Build, test and revise your tutor

**In this part:** you can say what a Claude Code skill is, have Claude plan and build one from your spec, check that it triggers, test it against the spec, and revise it. This is steps 4 and 5 from the table at the start of part 3.

You'll build the tutor inside this repository, at `.claude/skills/bf550-tutor/`, so that git sees every change Claude makes to it while you test and revise. The last step installs a copy in your home folder, so you can use it for the rest of the course.

### 4.1 · What a skill is

A **skill** is a folder holding one file, `SKILL.md`, that teaches Claude Code how to do one kind of task. The file opens with a short header:

```markdown
---
name: bf550-tutor
description: One or two sentences saying when to use this skill.
---
```

and continues with plain instructions in markdown. Claude Code reads every skill's `description` at the start of a session, and loads the rest of the file only when a request matches it. The folder's name becomes a command, so a skill in a folder named `bf550-tutor/` can also be started by typing `/bf550-tutor`.

Skills in `~/.claude/skills/` work in every folder you start Claude Code in. Skills in a repository's `.claude/skills/` work only there, and travel with the repository.

### 4.2 · Step 4: plan the build

Start a fresh conversation with `/clear`: everything Claude needs is in your spec now. Press **Shift+Tab** until the status line says *plan mode*. In plan mode Claude Code reads and plans, but changes no files. Then paste:

> Read `tutor-spec.md` and plan a Claude Code skill at `.claude/skills/bf550-tutor/SKILL.md` in this folder that implements it. Show me the description you'd write, under 200 characters and in the words I'd use to ask for help with BF550, and an outline of the instructions. For each part of the outline, say which section of my spec it comes from.

Read the plan against your spec before you accept it:

- [ ] every section of your spec shows up somewhere in the outline;
- [ ] nothing in the outline comes from nowhere: if it does, ask why, and keep it or cut it;
- [ ] the description uses words you'd actually type.

If something's wrong, say so and let Claude revise the plan. You can go round several times.

### 4.3 · Build it and read it

When the plan is right, accept it. Claude Code leaves plan mode and writes the file. Then read what it built: open `.claude/skills/bf550-tutor/SKILL.md` in the editor, with `tutor-spec.md` beside it. (`.claude` starts with a dot, so some file browsers hide it; VS Code shows it.) Is everything from the spec there? Is anything there that the spec didn't ask for?

Run `git status` in your second terminal: the skill is a new, untracked file. Commit this first version now, so that every change you make while testing shows up in `git diff`. In Claude Code, ask:

> Commit the skill as you built it, with a message that says so.

### 4.4 · Does it trigger?

Quit Claude Code (`/exit`) and start it again **in this folder**, so it reads the new skill. Try both ways in:

1. Type `/bf550-tutor` and answer its first question.
2. Without the command, ask for help with something from the course in your own words, for example "Can you help me understand odds ratios from the unit 10 lecture?" Did Claude use the skill? Ask it, if you are not sure.

A skill that never triggers is a description problem: Claude decides from the description alone.

### 4.5 · Step 5: test it against your spec

- **The tutor didn't do what the spec says.** Ask Claude Code to fix `SKILL.md` so it follows `tutor-spec.md`. Then run `git diff .claude/skills/` and read what it changed.
- **The tutor did what the spec says, and you don't like it.** You learned something about how you learn. Change `tutor-spec.md` first, then ask Claude Code to update the skill from it.
